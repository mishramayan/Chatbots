```python
## Message History
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

store={}

def get_session_history(session_id: str)->BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]

with_message_history = RunnableWithMessageHistory(model, get_session_history)
```

```python
config = {"configurable": {"session_id": "chat1"}}
```

```python
response = with_message_history.invoke(
    [
        HumanMessage(content="Hi, my name is Mayan and I'm a Senior Software Engineer.")
    ], config=config
)
```

```python
response.content
```

```python
## Prompt Templates
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder

prompt = ChatPromptTemplate.from_messages(
    [
        ("system", "You're a helpful assistant. Answer questions to the best of your ability in {language}"),
        MessagesPlaceholder(variable_name="messages")
    ]
)

chain = prompt | model
```

```python
response = chain.invoke({"messages": [HumanMessage(content="Hi my name is Mayan")], "language":"Hindi"})
response.content
```

```python
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
)
```

```python
config = {"configurable": {"session_id": "chat4"}}
response = with_message_history.invoke(
    {'messages': [HumanMessage(content="Hi, I am Mayan")], "language":"Spanish"},
    config = config
)
response.content
```

```python
from langchain_core.messages import SystemMessage, trim_messages

trimmer = trim_messages(
    max_tokens=40,
    strategy="last",
    token_counter=model,
    include_system=True,
    allow_partial=False,
    start_on="human"
)
messages = [
    SystemMessage(content="You're a helpful assistant. Answer questions to the best of your ability in {language}"),
    HumanMessage(content="Hi, I am Mayan"),
    AIMessage(content="Hi Mayan! Nice to meet you. How can I assist you today?"),
    HumanMessage(content="I like to eat ice cream and I like to play cricket. I also like to play football."),
    AIMessage(content="nice"),
    HumanMessage(content="What is 2 + 2?"),
    AIMessage(content="4"),
    HumanMessage(content="Thanks"),
    AIMessage(content="no problem"),
    HumanMessage(content="Having fun?"),
    AIMessage(content="yes"),
]

trimmer.invoke(messages)
```

```python
from operator import itemgetter
from langchain_core.runnables import RunnablePassthrough

chain = (
    RunnablePassthrough.assign(messages=itemgetter("messages") | trimmer)
    | prompt
    | model
)

response = chain.invoke(
    {
        "messages": messages + [HumanMessage(content="What math problem did I ask for?")],
        "language": "English"
    }
)
response.content
```
